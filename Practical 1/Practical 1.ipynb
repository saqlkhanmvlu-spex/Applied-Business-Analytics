```python
#Correlation Matrix
df.corr(numeric_only=True)
```

```python
#Histogram
plt.hist(df["JP_Sales"])
plt.title("Sales Revenue Distribution")
plt.xlabel("JP Sales Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
#Bar chart
df["Genre"].value_counts().plot(kind="bar")
plt.title("Genre")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
#Pie chart
df["Year"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Publishing Year")
plt.ylabel("")
plt.show()
```

```python
#Box plot
sns.boxplot(y=df["Global_Sales"])
plt.title("Global Sales Revenue Box Plot")
plt.show()
```

```python
#Correlation Heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="Blues")
plt.title("Correlation Heatmap")
plt.show()
```
